# 4. Enter Scores

Update the scores below after each team presentation.

Each score should be between **0 and 25**.


````python
scores = [
    {
        "team_id": 1,
        "team_name": "Team Alpha",
        "business_relevance": 0,
        "database_usage": 0,
        "architecture_clarity": 0,
        "demo_quality": 0,
        "judge_comments": "Add comments here.",
        "improvement_suggestions": "Add improvement suggestions here."
    },
    {
        "team_id": 2,
        "team_name": "Team Beta",
        "business_relevance": 0,
        "database_usage": 0,
        "architecture_clarity": 0,
        "demo_quality": 0,
        "judge_comments": "Add comments here.",
        "improvement_suggestions": "Add improvement suggestions here."
    },
    {
        "team_id": 3,
        "team_name": "Team Gamma",
        "business_relevance": 0,
        "database_usage": 0,
        "architecture_clarity": 0,
        "demo_quality": 0,
        "judge_comments": "Add comments here.",
        "improvement_suggestions": "Add improvement suggestions here."
    }
]

scores_df = pd.DataFrame(scores)
scores_df
````

# 5. Validate Scores

This checks that each score is between 0 and 25.


````python
score_columns = [
    "business_relevance",
    "database_usage",
    "architecture_clarity",
    "demo_quality"
]

def validate_scores(df):
    issues = []
    for col in score_columns:
        invalid_rows = df[(df[col] < 0) | (df[col] > 25)]
        if not invalid_rows.empty:
            issues.append(f"{col} has scores outside 0–25.")
    return "\n".join(issues) if issues else "All scores are valid."

validate_scores(scores_df)
````

# 6. Calculate Total Scores and Ranking


````python
results_df = scores_df.copy()
results_df["total_score"] = results_df[score_columns].sum(axis=1)
results_df["rank"] = results_df["total_score"].rank(method="dense", ascending=False).astype(int)

results_df = results_df.sort_values(
    by=["total_score", "business_relevance", "demo_quality"],
    ascending=False
)

results_df
````

# 7. Merge Team Details with Scores


````python
evaluation_summary_df = teams_df.merge(
    results_df,
    on=["team_id", "team_name"],
    how="left"
).sort_values(by="total_score", ascending=False)

evaluation_summary_df
````

# 8. Top Teams


````python
top_teams_df = evaluation_summary_df[
    evaluation_summary_df["rank"] == evaluation_summary_df["rank"].min()
]

top_teams_df
````

# 9. Award Suggestions

These are simple score-based suggestions. Judges can adjust final awards manually.


````python
def get_top_by_column(df, column):
    top_row = df.sort_values(by=column, ascending=False).iloc[0]
    return {
        "team_name": top_row["team_name"],
        "project_title": top_row.get("project_title", ""),
        "score": top_row[column]
    }

award_suggestions = {
    "Best Business Use Case": get_top_by_column(evaluation_summary_df, "business_relevance"),
    "Best Database Usage": get_top_by_column(evaluation_summary_df, "database_usage"),
    "Best AI Architecture": get_top_by_column(evaluation_summary_df, "architecture_clarity"),
    "Best Demo": get_top_by_column(evaluation_summary_df, "demo_quality"),
    "Best Overall Capstone": get_top_by_column(evaluation_summary_df, "total_score"),
}

award_suggestions
````

# 10. Feedback Report for Each Team


````python
def generate_team_feedback(row):
    return f'''
## Feedback for {row["team_name"]}

**Project:** {row["project_title"]}

**Target User:** {row["target_user"]}

**Architecture:** {row["architecture"]}

### Scores

- Business Relevance: {row["business_relevance"]}/25
- Database Usage: {row["database_usage"]}/25
- AI Architecture Clarity: {row["architecture_clarity"]}/25
- Demo Quality: {row["demo_quality"]}/25
- Total Score: {row["total_score"]}/100
- Rank: {row["rank"]}

### Judge Comments

{row["judge_comments"]}

### Improvement Suggestions

{row["improvement_suggestions"]}

---
'''

feedback_reports = [generate_team_feedback(row) for _, row in evaluation_summary_df.iterrows()]
print("\n".join(feedback_reports))
````

# 11. Generate Final Evaluation Report


````python
def dataframe_to_markdown(df):
    return df.to_markdown(index=False)

def generate_final_evaluation_report():
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M")

    award_text = "\n".join([
        f"- **{award}**: {details['team_name']} — {details['project_title']} ({details['score']})"
        for award, details in award_suggestions.items()
    ])

    feedback_text = "\n".join(feedback_reports)

    report = f'''
# Final Capstone Evaluation Report

Generated: {timestamp}

## Session

Zain Jordan Customer 360 AI Workshop  
Class 9: Final Capstone Presentations and Evaluation

---

## Evaluation Rubric

{dataframe_to_markdown(rubric)}

---

## Final Ranking

{dataframe_to_markdown(evaluation_summary_df[[
    "rank",
    "team_name",
    "project_title",
    "target_user",
    "business_relevance",
    "database_usage",
    "architecture_clarity",
    "demo_quality",
    "total_score"
]])}

---

## Award Suggestions

{award_text}

---

## Team Feedback

{feedback_text}

---

## General Next Steps

1. Improve data validation and edge-case handling.
2. Add clearer user interface if time allows.
3. Add logging and monitoring for tool usage.
4. Strengthen guardrails for production use.
5. Convert promising prototypes into production-ready architecture.
6. Consider MCP integration for reusable enterprise tools.
'''
    return report

final_report = generate_final_evaluation_report()
print(final_report[:4000])
````

# 12. Save Final Evaluation Report


````python
output_report_path = Path("final_capstone_evaluation_report.md")
output_report_path.write_text(final_report, encoding="utf-8")
print(f"Final evaluation report saved to: {output_report_path.resolve()}")
````
