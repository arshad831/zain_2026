# 5. Architecture Selection Guide

Use this guide to decide which AI pattern fits your project.

| Project Need | Best Pattern |
|---|---|
| Retrieve one customer's data | LangChain Tools or MCP Tools |
| Ask broad analytical questions | SQL Agent |
| Recommend plans, campaigns, or add-ons | RAG |
| Combine multiple specialist tasks | Multi-Agent |
| Expose tools in enterprise style | MCP |
| Draft customer messages | Care Message Agent |
| Build final integrated app | Multi-Agent + selected tools |

## Simple Rule

- If the question needs calculation → **SQL Agent**
- If the question needs explanation or recommendation → **RAG**
- If the question needs customer lookup → **Tool or MCP Tool**
- If the workflow has many steps → **Multi-Agent**


````python
architecture_guide = pd.DataFrame([
    {
        "Need": "Retrieve one customer’s profile, plan, billing, usage, or churn data",
        "Best Pattern": "LangChain Tool or MCP Tool",
        "Example": "Analyze customer 42"
    },
    {
        "Need": "Count, rank, group, compare, or summarize structured data",
        "Best Pattern": "SQL Agent",
        "Example": "Which city has the most high-risk churn customers?"
    },
    {
        "Need": "Recommend plans, add-ons, campaigns, or explain customer-experience themes",
        "Best Pattern": "RAG",
        "Example": "Which offer suits a heavy data user?"
    },
    {
        "Need": "Combine customer data, analysis, recommendation, and message writing",
        "Best Pattern": "Multi-Agent",
        "Example": "Analyze customer and recommend next best action"
    },
    {
        "Need": "Expose database tools in enterprise-style architecture",
        "Best Pattern": "MCP",
        "Example": "Expose get_customer_churn_risk as an MCP tool"
    },
    {
        "Need": "Write professional customer or internal messages",
        "Best Pattern": "Care Message Agent",
        "Example": "Draft a customer-care message"
    }
])

architecture_guide
````

# 20. Export Capstone Blueprint

This section creates a simple Markdown report from your team inputs.

You can download or copy the report for review.


````python
def list_to_markdown(items):
    return "\n".join([f"- {item}" for item in items])


def dataframe_to_markdown(df):
    return df.to_markdown(index=False)


def generate_capstone_blueprint():
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M")

    report = f'''
# Capstone Project Blueprint

Generated: {timestamp}

## 1. Team Information

**Team Name:** {team_info["team_name"]}

**Team Members:**
{list_to_markdown(team_info["team_members"])}

**Roles:**
- Business Lead: {team_info["business_lead"]}
- Data Lead: {team_info["data_lead"]}
- AI / Agent Lead: {team_info["ai_agent_lead"]}
- Demo Lead: {team_info["demo_lead"]}
- QA Lead: {team_info["qa_lead"]}

---

## 2. Project Overview

**Project Title:** {capstone_project["project_title"]}

**Business Problem:**  
{capstone_project["business_problem"]}

**Target User:**  
{capstone_project["target_user"]}

**Why This Project Matters:**  
{capstone_project["why_this_project_matters"]}

---

## 3. Main Demo Prompt

```text
{main_demo_prompt.strip()}
```

---

## 4. Database Tables

{list_to_markdown(database_tables)}

---

## 5. Table Mapping

{dataframe_to_markdown(table_mapping)}

---

## 6. AI Architecture Selection

- Use Python Functions: {architecture_selection["use_python_functions"]}
- Use LangChain Tools: {architecture_selection["use_langchain_tools"]}
- Use SQL Agent: {architecture_selection["use_sql_agent"]}
- Use RAG: {architecture_selection["use_rag"]}
- Use Multi-Agent: {architecture_selection["use_multi_agent"]}
- Use MCP: {architecture_selection["use_mcp"]}
- Use Care Message Agent: {architecture_selection["use_care_message_agent"]}
- Optional UI: {architecture_selection["optional_ui"]}

---

## 7. Agent / Tool Design

{dataframe_to_markdown(agent_tool_design)}

---

## 8. Expected Output Sections

{list_to_markdown(expected_output_sections)}

---

## 9. Demo Flow

{dataframe_to_markdown(demo_flow)}

---

## 10. Minimum Viable Demo

**Must Have:**
{list_to_markdown(minimum_viable_demo["must_have"])}

**Nice to Have:**
{list_to_markdown(minimum_viable_demo["nice_to_have"])}

**First Build Step:**  
{minimum_viable_demo["first_build_step"]}

**Second Build Step:**  
{minimum_viable_demo["second_build_step"]}

**Third Build Step:**  
{minimum_viable_demo["third_build_step"]}

---

## 11. Build Plan

{dataframe_to_markdown(build_plan)}

---

## 12. Evaluation Criteria

{dataframe_to_markdown(evaluation_criteria)}

---

## 13. Final Presentation Structure

{dataframe_to_markdown(presentation_structure)}
'''
    return report


blueprint_report = generate_capstone_blueprint()
print(blueprint_report[:3000])
````
